**Группировка по donor_id**

Далее будем группировать данные по donor_id, ведь наша модель машинного обучения будет предсказывать вероятность ухода в отток для каждого донора нащего датасета. Соответственно каждая запись нашего датасета на входе модели машинного обучения должна соответствовать уникальному donor_id.

```python
df_final['donor_id'].value_counts()
```

перед группировкой по donor_id оставим  в датафрейме только транзакции, имеющие статусы '3' и '5', т.е. завершенные либо отмененные т.к. они составляют почти 99% вcех транзакций. (Статус '-1' (около 1% транзакций) и статус '6' вообще не имеют описания, поэтому их удаляем из датасета. Статусы '1', '2' являются промежуточными и имеют менее 0.25%, статус 4 встречается менее чем в 0.1% случаев. Удаляем и их.)

```python
#Создадим для группировки новый отдельный датафрейм
df_grouped = df_final[(df_final['status'] == 3) | (df_final['status'] == 5)]
```

```python
# выделяем дату без времени из колонки 'when_x'
df_grouped['date'] = df_grouped['when_x'].dt.date
```

```python
# Удаляем колонку 'when_x'. Она нам больше не нужны
df_grouped.drop('when_x', axis=1, inplace=True)
```

```python
# преобразуем свежесозданную колонку с датой в тип данных 'дата'
df_grouped['date'] = pd.to_datetime(df_grouped['date'])
```

```python
# проверим соответствие типов данных
df_grouped.info()
```

```python
# создадим первый промежуточный датафрейм и вычислим значения полей для успешных транзакций (статус=3)
df_grouped_1 = (
    df_grouped[df_grouped['status'] == 3]
    .groupby('donor_id', as_index=False)
    .agg(
        sex=('sex', 'first'),  # Предположим, что пол одинаковый для одного donor_id
        sum_amount=('amount_total', 'sum'),  # сумма всех сделанных пожертований для каждого донора
        mean_amount=('amount_total', 'mean'),  # средний размер пожертования для каждого донора
        count_transactions_completed = ('amount_total', 'count'),   # количество успешных транзакций ( пожертвований)
        count_reccurent_transactions = ('is_reccurent', 'sum'),  # количсетво успешных реккурентных транзакций
        mean_date_diff = ('date', lambda dates: (dates.sort_values().diff().mean().days)),   # средняя разница в днях между успешными транзакциями
        first_transaction_date = ('date', 'min'),   # дата первой успешной транзакции донора
        last_transaction_date = ('date', 'max')     # дата последней успешной транзакции донора
    )
)
```

```python
# округлим значения в колонке 'mean_amount'
df_grouped_1['mean_amount'] = round(df_grouped_1['mean_amount'], 2)
```

```python
# создадим второй промежуточный датафрейм для подсчета количества отмененных транзакций (статус=5)
df_grouped_2 = (
    df_grouped[df_grouped['status'] == 5]
    .groupby('donor_id', as_index=False)
    .agg(
        count_transactions_cancelled = ('amount_total', 'count')
    )
)
```

```python
# объединяем 2 датафрейма в 1. ( Добавляем колонку с количествои отклоненных платежей к первому датафрейму)
df_grouped  = pd.merge(df_grouped_1, df_grouped_2, how='left', on='donor_id')
```

```python
df_grouped
```

```python
df_grouped.info()
```

```python
# заменяем значения NaN в колонке с количествои отмененных транзакций на 0.
df_grouped['count_transactions_cancelled'] = df_grouped['count_transactions_cancelled'].fillna(0).astype(int)
```

```python
# Добавим столбец с процентом отменненных транзакций
df_grouped['cancelled_percentage'] = 100 * round(df_grouped['count_transactions_cancelled'] / (df_grouped['count_transactions_completed'] + df_grouped['count_transactions_cancelled']), 3)
```

```python
# проверим результат
df_grouped.head()
```

```python
df_grouped['cancelled_percentage'].value_counts()
```

```python
# посмотрим наиболее "позднюю" дату в дашем датафрейме
df_grouped['last_transaction_date'].sort_values(ascending=False)
```

Мы видим, что данные у нас есть по 30.06.2021. т.е. по конец первого полугодия 2021 года. На сегодняшний день, очевидно, что данные являются устаревшими для реального прогноза оттока или даты следующего платежа по тому или иному донору. Собственно и отток, и следующие платежи уже прошли и остались в прошлом за прошедшие с момента сбора данных три с лишним года. Поэтому для подготовки нашей модели будем считать, что сегодня у нас 01.07.2021 г. и наши данные выгружены вот буквально только что.

Теперь нужно определиться с тем, каких доноров мы будем считать активными на данный момент. Тех, у кого есть хотя бы 2  успешных транзакции, сделанные не в 1 день
 ( а значит есть и интервал между ними, а соответственно, и средний интервал между успешными транзакциями) будем считать активными при условии, что с последней успешной транзакции по сегодняшнее число (01.07.2021) прошло времени не более среднего интервала между транзакциями для этого самого донора. Доноров у которых только 1 успешная транзакция, а значит нет никаких интервалов, будем считать активными, если с момента их единственной транзакции по сегодняшнее число (01.07.2021) прошло времени не более общего среднего значения интервала по всем донорам из нашего датафрейма df_grouped.

```python
# заменяем все NULL значения в колонке 'mean_date_diff' на нули
df_grouped['mean_date_diff'] = df_grouped['mean_date_diff'].fillna(0)
```

```python
# Вычисляем среднее количество дней между успешными транзакциями в целом по датафрейму.
#  Сумму всех средних интервалов, на количество доноров, чьи интервалы больше 0.
print(f'Среднее значение интервала между успешными транзакциями по всему датафрейму: {round(df_grouped['mean_date_diff'].sum()/ len(df_grouped[df_grouped['mean_date_diff'] > 0]))} дней')
```

```python
# заменим все нулевые средние интервалы на полученное выше значение
df_grouped['mean_date_diff'] = df_grouped['mean_date_diff'].replace(0, 166).astype(int)
```

```python
# Введем переменную для текущей даты
comparison_date = pd.to_datetime('2021-07-01')
```

```python
comparison_date
```

Добавляем колонку 'churn' которая фактически означает что донор не активен (отток) (1- не активен(отток), 0 - активен).  За отток донора принимаем условие, что разница между 01.07.2021 
 и днем последней транзакции донора больше, чем среднее значение интервала между платежами (mean_date_diff) данного донора. Эта колонка будет у нас целевой переменной при прогнозировании оттока с помощью машинного обучения

```python

df_grouped['churn'] = ((comparison_date - df_grouped['last_transaction_date']) > pd.to_timedelta(df_grouped['mean_date_diff'], unit='d')).astype(int)
```

```python
# Проверяем результат
df_grouped.head()
```

Добавим категориальную колонку 'has_reccure', которая означает является ли донор реккурентным подписчиком. 
Для этого будут учитываться только те доноры, которые делали реккуретные платежи и которые являются активными (churn=0)

```python

#df_grouped['has_reccure'] =  ((df_grouped['count_reccurent_transactions']) > 0).astype(int)
df_grouped['has_reccure'] = np.where((df_grouped['churn']==0) & (df_grouped['count_reccurent_transactions'] > 0), 1, 0)
```

```python
df_grouped.info()
```

```python
# Преобразуем все поля с датами в целые числа  с помощью функции toordinal для использования в качестве дискретного признака в модели машинного обучения
df_grouped['last_transaction_date'] = df_grouped['last_transaction_date'].apply(lambda x: x.date().toordinal())
df_grouped['first_transaction_date'] = df_grouped['first_transaction_date'].apply(lambda x: x.date().toordinal())
```

```python
# вводим dummy-переменные для категориальных признаков. В нашем датасете это только sex( Пол донора). Donor_id мы в состав признаков включать не будем.
dummies = pd.get_dummies(df_grouped['sex'], prefix='sex')
df_grouped = pd.concat([df_grouped, dummies], axis=1)
```

```python
# Конвертируем новые признаки из логических в целочисленные
df_grouped['sex_f'] = df_grouped['sex_f'].astype(int)
df_grouped['sex_m'] = df_grouped['sex_m'].astype(int)
df_grouped['sex_unknown'] = df_grouped['sex_unknown'].astype(int)
```

```python
# Удалим колонку 'sex'. После создания dummy-переменных для нее, она нам больше не нужна

df_grouped.drop('sex', axis=1, inplace=True)
```
